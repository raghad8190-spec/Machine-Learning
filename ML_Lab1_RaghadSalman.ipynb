```python
# Dataset
study_hours = [1, 2, 3, 4, 5, 6, 7]
marks = [50, 55, 60, 65, 72, 80, 88]

# 1. Number of samples
print("Number of samples:", len(marks))

# 2. Average study hours
print("Average study hours:", sum(study_hours) / len(study_hours))

# 3. Average marks
print("Average marks:", sum(marks) / len(marks))

# 4. Highest and lowest marks
print("Highest mark:", max(marks))
print("Lowest mark:", min(marks))

# 5. Display each sample
for i in range(len(study_hours)):
    print("Study hours:", study_hours[i], "Mark:", marks[i])

# 6. Pass/Fail
for i in range(len(marks)):
    if marks[i] >= 60:
        print("Mark:", marks[i], "Pass")
    else:
        print("Mark:", marks[i], "Fail")

# 7. Count the number of passes
passes = 0

for mark in marks:
    if mark >= 60:
        passes += 1

print("Number of passes:", passes)
```
